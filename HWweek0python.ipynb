```python
import requests
from pypdf import PdfReader
import re
from io import BytesIO
import time
session = requests.Session()
session.headers.update({
    "User-Agent": "Mozilla/5.0",
    "Accept": "application/json"
})
def get_org_id(inn):
    
    res = session.get(
        f"https://bo.nalog.gov.ru/advanced-search/organizations/search"
        f"?query={inn}&page=0"
    ).json()
    
    if not res["content"]:
        return None
    
    return res["content"][0]["id"]
def get_bfo(org_id):
    
    res = session.get(
        f"https://bo.nalog.gov.ru/nbo/organizations/{org_id}/bfo/"
    ).json()
    
    return res
def get_company_data(inn):
    
    org_id = get_org_id(inn)
    
    if org_id is None:
        return None
    
    bfo = get_bfo(org_id)
    
    return bfo
def get_report_text(org_id, year, correction):
    
    clarification_id = None
    
    if correction.get("clarification"):
        clarification_id = correction["clarification"].get("id")
    
    if clarification_id:
        
        url = f"https://bo.nalog.gov.ru/download/clarification/{clarification_id}"
        
        response = session.get(url, timeout=60)
        
        if response.status_code == 200 and response.content.startswith(b"%PDF"):
            
            pdf = PdfReader(
                BytesIO(response.content),
                strict=False
            )
            
            text = ""
            
            for page in pdf.pages:
                page_text = page.extract_text()
                if page_text:
                    text += page_text + "\n"
            
            return text
    
    correction_id = correction["id"]
    
    url = (
        f"https://bo.nalog.gov.ru/download/bfo/pdf/"
        f"{org_id}?period={year}&detailId={correction_id}"
    )
    
    response = session.get(url, timeout=60)
    
    if response.status_code == 200 and response.content.startswith(b"%PDF"):
        
        pdf = PdfReader(
            BytesIO(response.content),
            strict=False
        )
        
        text = ""
        
        for page in pdf.pages:
            page_text = page.extract_text()
            if page_text:
                text += page_text + "\n"
        
        return text
    
    return ""
def get_nma_amortization(text, year):

    lines = text.splitlines()

    start = None

    for i, line in enumerate(lines):
        if "наличие и движение нематериальных активов" in line.lower():
            start = i
            break

    if start is None:
        return None

    for line in lines[start:start + 150]:

        match = re.search(
            rf"за\s+{year}\s*г\.\s*(.*)",
            line,
            re.IGNORECASE
        )

        if not match:
            continue

        values_text = match.group(1)

        numbers = re.findall(
            r'\(?-?\d[\d\s]*(?:[,.]\d+)?\)?',
            values_text
        )

        if len(numbers) >= 3:

            value = numbers[2]

            value = value.replace(" ", "")
            value = value.replace(",", ".")

            if value.startswith("(") and value.endswith(")"):
                value = value[1:-1]

            try:
                return abs(float(value))
            except:
                return None

    return None
```

```python
def get_company_data(inn):

    org_id = get_org_id(inn)

    if org_id is None:
        return []

    bfo = get_bfo(org_id)

    results = []

    reports = {}

    for period_data in bfo:

        year = int(period_data["period"])

        annual_list = [
            x for x in period_data["typeCorrections"]
            if x["type"] == 12
        ]

        if not annual_list:
            continue

        correction = annual_list[0]["correction"]

        text = get_report_text(
            org_id,
            year,
            correction
        )

        reports[year] = {
            "correction": correction,
            "text": text
        }
    for year, report in reports.items():

        correction = report["correction"]
        text = report["text"]

        financial_result = correction["financialResult"]

        profit_before_tax = financial_result.get("current2300")
        interest_expense = financial_result.get("current2330")

        # Амортизация НМА из отчёта этого года
        amortization_nma = get_nma_amortization(
            text,
            year
        )
        if amortization_nma is None:

            next_year = year + 1

            if next_year in reports:

                next_text = reports[next_year]["text"]

                amortization_nma = get_nma_amortization(
                    next_text,
                    year
                )
        if (
            profit_before_tax is not None
            and interest_expense is not None
            and amortization_nma is not None
        ):

            ebita = (
                profit_before_tax
                + abs(interest_expense)
                + abs(amortization_nma)
            )

        else:
            ebita = None

        results.append({
            "ИНН": inn,
            "Год": year,
            "Прибыль до налогообложения": profit_before_tax,
            "Проценты к уплате": abs(interest_expense) if interest_expense is not None else None,
            "Амортизация НМА": amortization_nma,
            "EBITA": ebita
        })

    return results
```
